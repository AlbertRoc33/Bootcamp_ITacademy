Nivell 3.1
Crea una funció que prengui un dataframe com a paràmetre d'entrada.
La funció ha de crear (i exportar) un gràfic automàticament per a cada columna del dataframe. Per exemple:
un histograma/boxplot si la variable és numèrica
unes barres dels valors més freqüents si és categòrica
unes barres dels anys més freqüents si la dada està en format data.

```python
##Genera gràfics automàtics segons el tipus de columna:
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt

df_penguins = sns.load_dataset("penguins")

def genera_grafics_auto(df):
    for col in df.columns:
        s = df[col].dropna()  # treu valors nuls per evitar errors

        try:
            if pd.api.types.is_numeric_dtype(s):## 1ertipus: Numèrica
                s.plot(kind="hist", bins=20, title=f"Histograma de {col}", alpha=0.7) # Histograma
                plt.xlabel(col)
                plt.ylabel("Freqüència")
                plt.tight_layout()
                plt.show()

                s.plot(kind="box", title=f"Boxplot de {col}")  # Boxplot
                plt.tight_layout()
                plt.show()

            elif pd.api.types.is_datetime64_any_dtype(s):# Cas 2: Dates
                anys = s.dt.year.value_counts().sort_index()
                anys.plot(kind="bar", title=f"Freqüència per any ({col})")
                plt.xlabel("Any")
                plt.ylabel("Comptatge")
                plt.tight_layout()
                plt.show()

            else:
                counts = s.value_counts().head(10)  # Cas 3: Categòrica 
                counts.plot(kind="bar", title=f"Valors més freqüents de {col}")
                plt.xlabel(col)
                plt.ylabel("Comptatge")
                plt.tight_layout()
                plt.show()

        except Exception as e:
            print(f" No s'ha pogut generar el gràfic per {col}: {e}")
genera_grafics_auto (df_penguins)
```

Nivell 3 3.2 Carrega l'arxiu matriu_distancies.xlsx a pandas, de manera que els noms de files i els noms de columnes siguin els de les ciutats. Borra "Las Palmas de Gran Canaria" i "Palma" perquè poguem fer el trajecte en cotxe.
Ens interessa visitar totes les ciutats principals d'Espanya recorrent la mínima distància possible.
No cal que ho facis de forma òptima, ens interessa que desenvolupis una solució raonable utilitzant les eines que tens actualment.
Per exemple, una aproximació senzilla (que no òptima) seria anant sempre a la ciutat més propera que no haguem visitat encara
Fes una funció que donada la matriu de distàncies i la ciutat d'origen, faci una proposta de ruta que sigui el més curta possible que puguis, retornant una llista amb l'ordre de visita. Dóna també la distància total recorreguda.


```python
df4=pd.read_excel(r"C:\Users\Portátil\Desktop\Especialització_Data_Analytics\Sprint_10\matriu_distancies.xlsx",index_col=0)
```

```python
df4.head(4)
```

```python
df4.info()
```

```python
# canviem Nan per 0 que son els quilometres
df4=df4.fillna(0)
```

```python
#eliminem Palma // las palmas... seguiem en df4
df4.drop(columns=["Palma","Las Palmas de Gran Canaria"],inplace=True)
df4.drop(index=["Palma","Las Palmas de Gran Canaria"],inplace=True)
```

```python
df4.head(5)
```

```python
#des de la ciutat d'origen, sempre anem a la ciutat més propera no visitada.
def ruta_mes_curta_if(df4, origen):
     
    ciutats = list(df4.columns)# Llista de totes les ciutats
    visitades = [origen]# Inicialitzem ruta, segons on indiquem
    actual = origen
    dist_total = 0

    for _ in range(len(ciutats) - 1): # Iterem tantes vegades com ciutats queden per visitar
        no_visitades = []  # 1. Construir llista de ciutats no visitades
        for c in ciutats:
            if c not in visitades:
                no_visitades.append(c)

        distancies = df4.loc[actual, no_visitades] # 2. Obtenir distàncies des de la ciutat actual cap a les no visitades

        propera = None  # 3. Buscar la ciutat més propera amb if explícits
        min_dist = float("inf")
        for ciutat, dist in distancies.items():
            if dist < min_dist:
                min_dist = dist
                propera = ciutat

        visitades.append(propera)  # 4. Actualitzar ruta i distància total
        dist_total += min_dist
        actual = propera

    return visitades, dist_total

```

```python
#
ruta, distancia = ruta_mes_curta_if(df4, origen="Valencia")
print("Ruta recomanada:", ruta)
print("Distància total:", distancia, "km")
```
