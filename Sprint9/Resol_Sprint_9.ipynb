1. Calculadora de l'índex de massa corporal
Escriu una funció que calculi l'IMC ingressat per l'usuari/ària, és a dir, qui ho executi haurà d'ingressar aquestes dades.
Pots obtenir més informació del seu càlcul en: 🔗 Índice de masa corporal IMC que es y como se calcula.
La funció ha de classificar el resultat en les seves respectives categories.
Consell: Intenta validar les dades prèviament, perquè enviï un missatge d’advertència si les dades introduïdes 
per l’usuari/a no estan en el format adequat o no pren valors raonables.

```python
def Calcul_IMC():
    """Funció que demana quilograms i altura per calcular IMC"""
    try:
        kg = float(input("Introdueix el teu pes (kg): "))
        altura = float(input("Diguem la teva altura (cm): ")) / 100
        operacio = kg / (altura ** 2)
        # Mirem en quin grup classificar la persona segons l'IMC
        if operacio < 18.5:
            categoria = "Baix pes"
        elif 18.5 <= operacio <= 24.9:
            categoria = "Pes normal"
        elif 25 <= operacio <= 29.9:
            categoria = "Sobrepès"
        else:
            categoria = "Obesitat"

        # Retornem tant l'IMC com la categoria
        return f"El teu IMC és {operacio:.2f} → {categoria}"

    except ZeroDivisionError:
        return "Error: l'altura no pot ser 0."
    except ValueError:
        return "Error: introdueix només valors numèrics."

print(Calcul_IMC())
```

2. Convertidor de temperatures
Existeixen diverses unitats de temperatura utilitzades en diferents contextos i regions. Les més comunes són Celsius (°C), Fahrenheit (°F) i Kelvin (K). També existeixen altres unitats com Rankine (°Ra) i Réaumur (°Re).
Selecciona almenys 2 conversors, de tal manera que en introduir una temperatura retorni, com a mínim, dues conversions, de manera que es puguin guardar (recorda que un print() no es pot guardar mai).
Consell: Intenta validar les dades prèviament, perquè enviï un missatge d’advertència si les dades introduides per l’usuari/a no estan en el format adequat.


```python
def inicial_AZ(frase):
    """Retorna la inicial en minúscula si és lletra, o '#' si no."""
    if not frase:
        return "#"
    ch = frase[0]
    if "a" <= ch.lower() <= "z":
        return ch.lower()
    return "#"


def conversor_temperatura(tipus_temp: str, temperatura: float) -> dict:
    """
    Rep un tipus de temperatura ('celsius','fahrenheit','kelvin') i un valor numèric.
    Retorna un diccionari amb les conversions a les altres escales.
    """
    opcions = {"celsius", "fahrenheit", "kelvin"}
    tipus = tipus_temp.lower()

    if tipus not in opcions:
        print("Error: opcions vàlides = Celsius, Fahrenheit, Kelvin")
        return {}

    print("Has triat:", tipus)
    print("Has posat graus:", temperatura)

    Conversor = {}
    lletra_inicial = inicial_AZ(tipus)

    match lletra_inicial:
        case "c":
            cal_temp = temperatura * 9/5 + 32
            cal_temp2 = temperatura + 273.15
            Conversor["Fahrenheit"] = cal_temp
            Conversor["Kelvin"] = cal_temp2
        case "f":
            cal_temp3 = (temperatura - 32) / 1.8
            cal_temp4 = (temperatura - 32) / 1.8 + 273.15
            Conversor["Celsius"] = cal_temp3
            Conversor["Kelvin"] = cal_temp4
        case "k":
            cal_temp5 = temperatura - 273.15
            cal_temp6 = (temperatura - 273.15) * 9/5 + 32
            Conversor["Celsius"] = cal_temp5
            Conversor["Fahrenheit"] = cal_temp6

    return Conversor
# Demanar dades a l’usuari
tipus = input("Diguem el tipus de temperatura: Celsius, Fahrenheit, Kelvin ")
graus = float(input("Diguem els graus: "))

resultat = conversor_temperatura(tipus, graus)
print("Resultats de la conversió:", resultat)

```

1. Comptador i endreçador de paraules d'un text.
El client va quedar content amb el comptador de paraules, però ara vol llegir arxius TXT i que calculi la freqüència de cada paraula ordenades dins de les entrades habituals del diccionari segons la lletra amb la qual comencen, és a dir, les claus han d'anar de la A a la Z i dins de la A hem d'anar de la A la Z. Per exemple, per a l'arxiu "tu_me_quieres_blanca.txt" la sortida esperada seria:


```python
#  unim neteja amb classificacio.
def comptador_i_endresador_per_inicial(nom_fitxer):
    def neteja_frase(frase):
        frase = frase.lower()
        chars = ".,;:!?\"'()[]{}«»¡¿/\\—–-"
        return frase.strip(chars)

    def inicial_AZ(frase):
        if not frase:
            return "#"
        ch = frase[0]    ###Primer caracter de la paraula
        if "a" <= ch <= "z":  ### comprovació que està entre a i z minúscula
            return ch  ###Si és lletra, retorna la inicial
        return "#"

    grups = {}
    with open(nom_fitxer, "r", encoding="utf-8") as f:
        for linia in f:
            for part in linia.split():
                frase = neteja_frase(part)
                if not frase:
                    continue
                ini = inicial_AZ(frase)
                if ini not in grups:
                    grups[ini] = {}
                # --- Aquí fem servir el patró if/else ---
                if frase in grups[ini]:
                    grups[ini][frase] += 1
                else:
                    grups[ini][frase] = 1

    resultado = {}
    for lletra in sorted(grups.keys()): ###Recorrem les inicials en ordre alfabètic
        paraules_ordenades={} ###diccionari temporal per a les paraules d'aquesta inicial
        for paraula in sorted(grups[lletra].keys()): ###Recorrem les paraules de la inicial, també ordenades alfabèticament
            freq= grups[lletra][paraula] # agafem la freqüència de la paraula
            paraules_ordenades[paraula]= freq #la guardem al diccionari temporal
        
        resultado[lletra] =paraules_ordenades #assignem el diccionari temporal al resultat sota la inicial corresponent
    return resultado

def imprimeix_grups_ordenats(resultado):
    for lletra in sorted(resultado.keys()):
        print(f"[{lletra.upper()}]")
        for paraula, freq in resultado[lletra].items():
            print(f"  {paraula}: {freq}")
        print()

### Aquí cridem les funcions
resultado = comptador_i_endresador_per_inicial("tu me quieres blanca.txt")
imprimeix_grups_ordenats(resultado)
```

2. Conversió de tipus de dades.
El client rep una llista de dades i necessita generar dues llistes, la primera on estaran tots els elements que es van poder convertir en flotants i l'altra on estan els elements que no es van poder convertir. Exemple de la llista que rep el client:
[ '1.3', 'one' , '1e10' , 'seven', '3-1/2', ('2',1,1.4,'not-a-number'), [1,2,'3','3.4']]


```python
llista_pp = ['1.3', 'one', '1e10', 'seven', '3-1/2',
             ('2', 1, 1.4, 'not-a-number'),
             [1, 2, '3', '3.4']]

llista_float = []
llista_NO_float = []

for dades in llista_pp:
    # Si és llista o tupla, recórrer sub-elements directament
    if type(dades) in (list, tuple):
        for subdades in dades:
            try:
                valor = float(subdades)
                llista_float.append(valor)
            except (ValueError, TypeError):
                llista_NO_float.append(subdades)
    else:
        # Si és un valor simple, intentem convertir-lo
        try:
            valor = float(dades)
            llista_float.append(valor)
        except (ValueError, TypeError):
            llista_NO_float.append(dades)

print("Convertibles:", llista_float)
print("No convertibles:", llista_NO_float)
```

Explora el funcionament del mòdul random de la llibreria numpy. 🔗 Random module in NumPy
En aquest punt, el client ha detectat un problema amb les contrasenyes que utilitzen els seus treballadors. Asdf1234, dates d’aniversari o similars. Per a solucionar-ho ens ha encarregat una funció de Python que generi contrasenyes més segures. La funció ha de dependre dels següents paràmetres:
________________________________________
•	longitud (int): Longitud de la contrasenya
•	majuscules (bool = True): Si hi ha d’aparèixer majúscules
•	minuscules (bool = True): Si hi ha d’aparèixer minúscules
•	numeros (bool = True): Si hi ha d’aparèixer números
•	signes (bool = False): Si hi ha d’aparèixer caràcters especials (,-$? o similars)
________________________________________
Així doncs, si executem la funció de la següent manera:
crear_contrasenya(10, True, True, True, True)
Hauríem d’obtenir un output (que hem de poder guardar) de l’estil:
9Er,5Vn8P$
Assegura’t que es compleixin tots els criteris, i que aquestes contrasenyes siguin realment aleatòries.


```python

import numpy as np

def crear_contrasenya(longitud,majuscules=True,minuscules=True,numeros=True,signes=False):
    try:
        # 0) Construcció dels grups de caràcters de forma eficient
        grups = [] #llista on guardem caràcter seleccionats
        if majuscules:
            majuscules_list = list("ABCDEFGHIJKLMNOPQRSTUVWXYZ")
            grups.append(majuscules_list)
        if minuscules:
            minuscules_list = list("abcdefghijklmnopqrstuvwxyz")
            grups.append(minuscules_list)
        if numeros:
            numeros_list = list("0123456789")
            grups.append(numeros_list)
        if signes:
            signes_list = list("!@#$%^&*?-_")
            grups.append(signes_list)
        # Validacions
        if not grups:
            print("[Error] Has de seleccionar almenys un tipus de caràcter.")
            return None
        if longitud < len(grups):
            print("[Error] La longitud ha de ser com a mínim el nombre de grups seleccionats.")
            return None
        # 1) Garantim almenys un caràcter de cada grup
        contrasenya_chars = []
        for grup in grups:
            lletra = np.random.choice(grup)
            contrasenya_chars.append(lletra)
        cataleg = []  ## 2) Catàleg combinat
        for grup in grups:
            cataleg.extend(grup)
        nombre_faltants = longitud - len(contrasenya_chars) # 3) Completem fins a la longitud
        for i in range(nombre_faltants):
            lletra = np.random.choice(cataleg)
            contrasenya_chars.append(lletra)
        # 4) Barregem l’ordre amb NumPy
        contrasenya_chars = list(np.random.permutation(contrasenya_chars))
        # 5) Construïm la contrasenya
        contrasenya = "".join(contrasenya_chars)

        return contrasenya

    except Exception as e:
        print(f"[Error inesperat] {e}")
        return None

# Exemple d’ús
print(crear_contrasenya(10, True, True, True, True))
```


•	2. Processament de dades simple
Una companya de feina ens ha demanat un favor, aprofitant que sap que estem aprenent a programar. Té un històric de partits de futbol català en un fitxer, on hi ha emmagatzemat els noms dels equips i els resultats. Necessita que processem les dades de manera automàtica, per extreure’n els resultats que necessita.
Utilitza el fitxer "historic_partits.txt"
Necessita un programa que retorni:
________________________________________
•	El nombre total de gols que ha fet cada equip.
•	El nom de l’equip més golejador.
•	El nom de l’equip més golejat
•	La classificació global (cada victòria: 3 pts, empat 1 pts, derrota 0 pts)


```python
def net_frase(frase):
    fora_gui = frase.split("-")
    part1 = fora_gui[0].split()
    part2 = fora_gui[1].split()

    club1 = " ".join(part1[:-1])
    gols1 = int(part1[-1])

    club2 = " ".join(part2[1:])
    gols2 = int(part2[0])

    return [(club1, gols1, gols2), (club2, gols2, gols1)]


def processar_fitxer(nom_fitxer):
    gols_fets = {}
    gols_rebuts = {}
    punts = {}
    vict = {}
    emp = {}
    der = {}

    with open(nom_fitxer, "r", encoding="utf-8") as f:
        for linia in f:
            linia = linia.strip()
            if not linia:
                continue
            equips = net_frase(linia)

            club_local, gL, gV = equips[0]
            club_visitant, gV2, gL2 = equips[1]

            # Actualitzar gols fets i rebuts
            for club, gf, gr in equips:
                if club in gols_fets:
                    gols_fets[club] += gf
                else:
                    gols_fets.update({club: gf})

                if club in gols_rebuts:
                    gols_rebuts[club] += gr
                else:
                    gols_rebuts.update({club: gr})

            # Inicialitzar comptadors
            for club in [club_local, club_visitant]:
                if club not in punts: punts[club] = 0
                if club not in vict: vict[club] = 0
                if club not in emp: emp[club] = 0
                if club not in der: der[club] = 0

            # Actualitzar punts i estadístiques
            if gL > gV:  # victòria local
                punts[club_local] += 3
                vict[club_local] += 1
                der[club_visitant] += 1
            elif gL < gV:  # victòria visitant
                punts[club_visitant] += 3
                vict[club_visitant] += 1
                der[club_local] += 1
            else:  # empat
                punts[club_local] += 1
                punts[club_visitant] += 1
                emp[club_local] += 1
                emp[club_visitant] += 1

    return gols_fets, gols_rebuts, punts, vict, emp, der


# Funció auxiliar per ordenar per punts
def obtenir_punts(item):
    return item[1]


def informe(nom_fitxer):
    gols_fets, gols_rebuts, punts, vict, emp, der = processar_fitxer(nom_fitxer)

    equip_mes_golejador = max(gols_fets, key=gols_fets.get)
    equip_mes_golejat = max(gols_rebuts, key=gols_rebuts.get)

    classificacio = sorted(punts.items(), key=obtenir_punts, reverse=True)

    print("\n Taula completa de classificació i estadístiques:")

    # Construir capçalera amb llista + join
    headers = [
        "Club".ljust(15),
        "Pts".rjust(3),
        "W".rjust(3),
        "D".rjust(3),
        "L".rjust(3),
        "GF".rjust(4),
        "GA".rjust(4)
    ]
    print(" ".join(headers))
    print("-" * 50)

    # Files de dades
    for club, pts in classificacio:
        row = [
            club.ljust(15),
            str(pts).rjust(3),
            str(vict[club]).rjust(3),
            str(emp[club]).rjust(3),
            str(der[club]).rjust(3),
            str(gols_fets[club]).rjust(4),
            str(gols_rebuts[club]).rjust(4)
        ]
        print(" ".join(row))

    print("\n Equip més golejador:", equip_mes_golejador, gols_fets[equip_mes_golejador])
    print(" Equip més golejat:", equip_mes_golejat, gols_rebuts[equip_mes_golejat])


# Punt d’entrada
if __name__ == "__main__":
    informe("historic partits.txt")
```
